# Interactive PyTorch Face Attribute Inference

Loads deep learning checkpoints (Residual Conv2D CNN, Vision Transformer, Spatial BiLSTM, ANN) and runs forward passes with Test-Time Augmentation (TTA).


```python
# Cell 1: Load inference engine and check device
import os
import sys
from pathlib import Path

import numpy as np
import torch
from PIL import Image

PROJECT_ROOT = Path.cwd().parent.parent if Path.cwd().name in ['training', 'analysis', 'inference', 'verification'] else Path.cwd()
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from src.models.inference import ModelInferenceEngine

engine = ModelInferenceEngine()
print(f"Inference engine ready on device: {engine.device}")
```

```python
# Cell 2: Run inference on sample face image with and without TTA
# Create synthetic test image or load local CelebA image
sample_path = PROJECT_ROOT / "img_align_celeba" / "000001.jpg"
if sample_path.exists():
    test_input = str(sample_path)
    print(f"Testing on real CelebA image: {sample_path.name}")
else:
    test_input = np.random.randint(0, 255, (64, 64, 3), dtype=np.uint8)
    print("Testing on synthetic tensor.")

for arch in ["Residual CNN", "Vision Transformer", "Spatial BiLSTM", "ANN"]:
    res_standard = engine.predict_attribute(test_input, architecture=arch, use_tta=False)
    res_tta = engine.predict_attribute(test_input, architecture=arch, use_tta=True)
    print(f"{arch:<20} | Standard P: {res_standard['probability']:.4f} | TTA P: {res_tta['probability']:.4f} | Confidence: {res_tta['confidence'] * 100:.1f}%")
```
