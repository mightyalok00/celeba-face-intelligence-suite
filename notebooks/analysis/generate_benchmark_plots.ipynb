# Publication-Quality Benchmark Plot Generation

Generates comparative bar charts, confusion matrices, and ROC curves from evaluated model benchmark results.


```python
# Cell 1: Import visualization libraries and configure style
import os
import sys
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

PROJECT_ROOT = Path.cwd().parent.parent if Path.cwd().name in ['training', 'analysis', 'inference', 'verification'] else Path.cwd()
csv_path = PROJECT_ROOT / "reports" / "model_benchmark_results.csv"
df = pd.read_csv(csv_path)
print(f"Loaded {len(df)} model benchmarks from {csv_path.name}")
df.head()
```

```python
# Cell 2: Plot comparative bar charts across all architectures
fig, ax = plt.subplots(figsize=(12, 6))
models = df["Model Configuration"]

# Parse percentage strings if needed
if df["Accuracy"].dtype == object:
    acc_vals = df["Accuracy"].str.rstrip("%").astype(float)
else:
    acc_vals = df["Accuracy"] * 100.0

bars = ax.barh(models, acc_vals, color="#3b82f6", edgecolor="#1d4ed8", alpha=0.85)
ax.set_xlabel("Accuracy (%)", fontsize=12, fontweight="bold")
ax.set_title("Multi-Architecture Model Performance Comparison (CelebA)", fontsize=14, fontweight="bold", pad=15)
ax.set_xlim(70, 100)
ax.grid(axis="x", linestyle="--", alpha=0.5)

# Add value labels
for bar in bars:
    w = bar.get_width()
    ax.text(w + 0.5, bar.get_y() + bar.get_height() / 2, f"{w:.2f}%", va="center", fontsize=10)

plt.tight_layout()
plt.show()
```
